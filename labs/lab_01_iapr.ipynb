## Part 1.4 - Region Growing (4 pts)

* **Q1 (3 pts)** In this section we ask you to implement from scratch your own `region_growing` function to detect the tumor area. The function should take as input an RGB image, seeds that are manual initialization points in the tumor area, and the maximum number of iterations to perform. ote that the function also includes the kwargs argument. You can use it to add additional parameters to your function if needed (see [doc kwargs](https://book.pythontips.com/en/latest/args_and_kwargs.html)). Importantly, for timing reasons, a single explicit for loop on iterations is expected. An explicit for loop on pixels or neighbours will be accepted but penalised by 1 point. Use numpy tools to succeed.
* **Q2 (1 pts)** Your function is then used to perform region growing using multiple iterations. Comment on the quality of the results based on the number of iterations performed as well as the running time.

````python
 
def similar(hsv_values, hsv_values_seed):
    # define thresholds
    # H_th = 0.905
    # S_th = 0.295
    # V_th = 0.505
    h1, s1, v1 = hsv_values_seed
    # get the HSV values
    h, s, v = hsv_values
    # apply threshold to evaluate similarity True or False
    tolerance = 0.3
    return np.abs(h - h1) < tolerance and np.abs(s - s1) < tolerance and np.abs(v - v1) < tolerance

def get_neighbors(img, x_seed, y_seed):
    #  neighbors_region = 8
    neighbors = []
    for dx in range(-1, 2):
        for dy in range(-1, 2):
            if dx == 0 and dy == 0:
                continue
            nx, ny = x_seed + dx, y_seed + dy
            if 0 <= nx < img.shape[0] and 0 <= ny < img.shape[1]:
                neighbors.append((nx, ny))
    return neighbors

def rg_recursive(img, res, x, y, x_seed, y_seed, label):
    if x<0 or x>=img.shape[0] or y<0 or y>=img.shape[1]: # out of bounds
        return
    
    if res[x, y] != 0: # already visited
        return
    
    neighbors = get_neighbors(img, x, y)
    res[x, y] = label
    for nx, ny in neighbors:
        if similar(img[x_seed, y_seed], img[nx, ny]):
            res[nx, ny] = label
            rg_recursive(img, res, nx, ny, label)
    
    return res
    

def region_growing(
    seeds: list[tuple],
    img: np.ndarray,
    n_max: int = 10,
    **kwargs
):
    """
    Run region growing on input image using seed points.

    Args
    ----
    seeds: list of tuple
        List of seed points
    img: np.ndarray (M, N, C)
        RGB image of size M, N, C
    n_max: int
        Number maximum of iterations before stopping algorithm

    Return
    ------
    rg: np.ndarray (M, N)
        Image after region growing has been performed
    """
    
    M, N, _ = img.shape
    #rg = np.zeros((M, N)).astype(bool)
    rg = np.zeros((M, N), dtype=int)
    
    # ------------------
    # PSEUDOCODE:
    # (0) Convert to HSV
    img_hsv = rgb2hsv(img)
    # (1) Selection of a seed s, ie, an initial pixel. 
    label = 1
    for s in seeds:
        x, y = s
        # Get neighbors of seed
        rg_recursive(img_hsv, rg, x, y, label)
        label += 1

    # Aggregate  every pixel with label different from zero to be True and evcery pixel with label 0 to False
    rg = rg != 0

    # ------------------                 
    return rg
````

````python
import numpy as np
# !pip install opencv-python -q
import cv2
from skimage.color import rgb2hsv
from collections import deque
import matplotlib.pyplot as plt

def similar(hsv_values, hsv_seed, tolerance=0.5):
    """Check if pixel is similar to the seed pixel using HSV thresholds."""
    h, s, v = hsv_values
    h1, s1, v1 = hsv_seed
    return abs(h - h1) < tolerance and abs(s - s1) < tolerance and abs(v - v1) < tolerance

def get_neighbors(x, y, shape):
    """Get 8-connected neighbors of (x, y) within image bounds."""
    neighbors = [
        (x-1, y-1), (x-1, y), (x-1, y+1),
        (x, y-1),             (x, y+1),
        (x+1, y-1), (x+1, y), (x+1, y+1),
    ]
    return [(nx, ny) for nx, ny in neighbors if 0 <= nx < shape[0] and 0 <= ny < shape[1]]

def region_growing(img, seeds, tolerance=0.1):
    """
    Perform region growing segmentation.

    Args:
        img (np.ndarray): RGB image.
        seeds (list of tuple): List of seed points [(x, y), ...].
        tolerance (float): Threshold for similarity in HSV space.

    Returns:
        np.ndarray: Binary mask of segmented region.
    """
    # Convert image to HSV
    img_hsv = rgb2hsv(img)
    
    # Initialize mask
    mask = np.zeros(img.shape[:2], dtype=bool)
    
    # Initialize queue for BFS
    queue = deque(seeds)

    # Process each seed
    for x, y in seeds:
        mask[x, y] = True

    while queue:
        x, y = queue.popleft()  # Get pixel from queue
        seed_hsv = img_hsv[x, y]  # HSV values of seed pixel

        # Check neighbors
        for nx, ny in get_neighbors(x, y, img.shape[:2]):
            if not mask[nx, ny]:  # If not visited
                if similar(img_hsv[nx, ny], seed_hsv, tolerance):
                    mask[nx, ny] = True
                    queue.append((nx, ny))

    return mask  # Binary mask

# Load an image
img = img_he
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)  # Convert from BGR (OpenCV default) to RGB

# Set seed points (manually selected)
seeds = [(100, 150), (200, 250)]  # Example seed points

# Run region growing
result_mask = region_growing(img, seeds, tolerance=0.1)

# Display results
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(img)
plt.title("Original Image")

plt.subplot(1, 2, 2)
plt.imshow(result_mask, cmap="gray")
plt.title("Segmented Region")

plt.show()

````
